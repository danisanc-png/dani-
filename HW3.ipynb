```python
# Import packages 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# read the file
fileName = "/Users/danisanchez/Downloads/LA_AQS_2023.csv" #note on path
pd.read_csv(fileName)
```

```python
df = pd.read_csv(fileName) #Read it in and store it in a variable!
```

```python
df.head() # describe how the file looks
# We want a specific parameter Name and Duration Description
# Arithmetic Mean is what we love

#GOTO Filtering
```

```python
df['Parameter Name']
```

```python
df_O3 = df[(df['Parameter Name']=='Ozone') & (df['Duration Description']=='1 HOUR')]
df_O3.head()

#GOTO Plotting
```

```python
ax1 = df_O3.plot.scatter(x='Date (Local)',y='Arithmetic Mean') # Terrible x axis. 
```

```python
# Create a new column for daily average in ppbv
df_O3['Arithmetic Mean (ppbv)'] = df_O3['Arithmetic Mean'] * 1000
```

```python
sns.scatterplot(
    data=df_O3,
    x= df_O3['Date (Local)'],
    y=df_O3['Arithmetic Mean (ppbv)'],
    color= 'red',
    alpha=0.6,
    s=200 )
plt.show()
```

```python
# a. When is O3 highest?
highest_day = df_O3.loc[df_O3['Arithmetic Mean (ppbv)'].idxmax()]
print(f"Highest O3: {highest_day['Arithmetic Mean (ppbv)']} ppbv")
print(f"Date: {highest_day['Date (Local)']}")

# b. How many days are above 35 ppbv? (Using the daily peak hour)
days_above_35 = df_O3[df_O3['Arithmetic Mean (ppbv)'] > 35]
print(f"Number of days above 35 ppbv: {len(days_above_35)}")
```
